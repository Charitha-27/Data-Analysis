```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_excel('/content/population.xlsx')
df
```

```python
df.info()
```

# DATA CLEANING

```python

df.drop_duplicates(inplace=True)
df.dropna(inplace=True)
df.reset_index(inplace=True)
df.drop('index',axis=1,inplace=True)
#df.drop('level_0',axis=1,inplace=True)
df['LocID']=df['LocID'].astype(int)
df
```

```python
df['Time']=df['Time'].astype(int)
df.info()
```

```python
df['AgeGrpStart']=df['AgeGrpStart'].astype(int)
df.info()
```

```python
df['AgeGrpSpan']=df['AgeGrpSpan'].astype(int)
df.info()
```

```python
df['PopMale']=df['PopMale'].str.strip('ERROR_')
df['PopMale']=df['PopMale'].astype(float)
df['PopMale']=df['PopMale'].astype(int)
df['PopFemale']=df['PopFemale'].astype(int)
df.info()
```

# DATA VISUALIZATION

```python
loc=df.groupby('Location')['PopTotal'].sum().sort_values(ascending=False)
loc=pd.DataFrame(loc).iloc[:7,:]
plt.figure(figsize=(12,5))
a=sns.barplot(x='Location',y='PopTotal',data=loc,hue='Location',palette='plasma',errorbar=None)
a.set_xlabel('Location',fontsize=13,color='green')
a.set_ylabel('Total Population',fontsize=13,color='green')
a.set_title('Analysing the Total Population based on Location',color='red',fontstyle='oblique',fontsize=20)
a.set_xticklabels(a.get_xticklabels(),rotation=45)

def data_label(a):
  for i in a.containers:
    a.bar_label(i)

data_label(a)
plt.show()
```

```python
pop=df.pivot_table(values=['PopMale','PopFemale'],index='Location',aggfunc=sum)
pop.sort_values(by=['PopMale','PopFemale'],ascending=False,inplace=True)
pop=pop.iloc[:7,:]
fig,axes=plt.subplots(nrows=1,ncols=2,figsize=(25,5))
a=sns.barplot(x='Location',y='PopMale',data=pop,hue='Location',ax=axes[0],palette='Set2')
b=sns.barplot(x='Location',y='PopFemale',data=pop,hue='Location',ax=axes[1],palette='Set2')
axes[0].set_xlabel('Location',color='blue',fontsize=15)
axes[0].set_ylabel('Male Population',color='blue',fontsize=15)
axes[0].set_title('Analysing Total Popluation of Male based on top 7 locations',color='red',fontstyle='oblique',fontsize=20)
axes[1].set_xlabel('Location',color='blue',fontsize=15)
axes[1].set_ylabel('Female Popultion',color='blue',fontsize=15)
axes[1].set_title('Analysing Total Population of Female based on top 7 locations',color='red',fontstyle='oblique',fontsize=20)
axes[0].set_xticklabels(axes[0].get_xticklabels(),rotation=70)
axes[1].set_xticklabels(axes[1].get_xticklabels(),rotation=70)
data_label(a)
data_label(b)
plt.show()
```

```python
age=df.groupby('AgeGrp')['PopTotal'].sum().sort_values(ascending=False)
age=pd.DataFrame(age).iloc[:8:]
plt.pie(age['PopTotal'],autopct='%0.00f%%',labels=['0-4','5-9','10-14','15-19','20-24','25-29','30-34','35-39'])
plt.title('Analysing the Total Population based on Age Group',color='red',fontstyle='oblique')
plt.show()
```

```python
year=df.groupby('Time')['PopTotal'].sum()
year=pd.DataFrame(year).reset_index()
plt.figure(figsize=(13,5))
a=sns.lineplot(x='Time',y='PopTotal',data=year,marker='o',markerfacecolor='green',markeredgecolor='red',markersize=9)
a.set_xlabel('Year',color='blue',fontsize=15)
a.set_ylabel('Total Population',color='blue',fontsize=15)
a.set_title('Analysing the Total Population based on Year',color='red',fontstyle='oblique',fontsize=20)
for i,j in zip(year['Time'],year['PopTotal']):
  plt.annotate(f'{i},{j}',(i,j),ha='left')
plt.show()
```
